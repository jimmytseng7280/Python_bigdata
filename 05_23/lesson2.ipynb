```python
while True:
    try:
        a = int(input("請輸入起始值(整數)："))
        break
    except ValueError:
        print("輸入錯誤，請輸入整數！")

while True:
    try:
        b = int(input("請輸入結束值(整數)："))
        break
    except ValueError:
        print("輸入錯誤，請輸入整數！")

total = 0

for i in range(a,b+1):
    total += i
print("區間總和 = " + str(total))
```

```python
#九九乘法表
for i in range(1,10):
    for j in range(1,10):
        print(f"{i} x {j} = {i*j}")
    print()
```

```python
#成績等級（if多項選擇）

score = int(input("請輸入分數（0～100）："))

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
elif score >= 60:
    print("D")
else:
    print("不及格")
```
