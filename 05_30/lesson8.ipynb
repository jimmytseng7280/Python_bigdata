```python
file = open("student.txt", "r", encoding="utf-8")
print(type(file))
content = file.read()
print(content)
file.close()
file.closed
```

```python
#file.read()是實體方法:
#with open()是上下文管理器:會自動幫我們關閉檔案

with open("student.txt", "r", encoding="utf-8") as file:
    content = file.read()
    print(content)

file.closed
```
