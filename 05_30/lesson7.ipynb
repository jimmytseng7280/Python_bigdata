```python
try:
	with open("student.txt", "r", encoding="utf-8") as file:
			data = file.read()
			print(data)
except Exception as e:
	print("發生錯誤:", e)
```

```python
#open(flie="student.txt")引數名稱的呼叫

try:
	with open(file="student.txt", encoding="utf-8") as file:
			data = file.read()
			print(data)
except Exception as e:
	print("發生錯誤:", e)
```

```python
#open引數值呼叫(必須依順序呼叫)

try:
	with open("student.txt", 'r', -1 , 'utf-8') as file:
			data = file.read()
			print(data)
except Exception as e:
	print("發生錯誤:", e)
```

```python
#open引數名稱呼叫(可以不依順序呼叫)

try:
	with open(file="student.txt", encoding='utf-8', mode='r') as file:
			data = file.read()
			print(data)
except Exception as e:
	print("發生錯誤:", e)
```

```python
#open混合呼叫(前面使引數值依序呼叫，後面使用引數名稱呼叫，使用引數名稱後，後面都要使用引數名稱呼叫)

try:
	with open("student.txt", 'r', encoding='utf-8') as file:
			data = file.read()
			print(data)
except Exception as e:
	print("發生錯誤:", e)
```

```python
#file.name是實體屬性:可以取得檔案名稱
#file.read()是實體方法:可以一次讀取整個檔案內容，並以字串形式回傳

file = open("student.txt", "r", encoding="utf-8")

print(file.name) #實體屬性:可以取得檔案名稱，例如'student.txt'
print(file.mode) #實體屬性:可以取得檔案的開啟模式，例如'r'表示讀取模式，'w'表示寫入模式，'a'表示附加模式等
print(file.encoding) #實體屬性:可以取得檔案的編碼方式
print(file.closed) #實體屬性:可以檢查檔案是否已經關閉，返回True或False
print(file.read()) #實體方法:可以一次讀取整個檔案內容，並以字串形式回傳
file.close() #實體方法:可以關閉檔案，釋放系統資源
```
