```python
import numpy as np
from numpy import ndarray
import pandas as pd
from pandas import DataFrame

#scores:ndarray | None = None # 學生的分數
#scores_df:DataFrame | None = None # 學生分數的DataFrame

scores:ndarray = np.random.randint(40, 101, size=(50, 5))
scores_df:DataFrame = pd.DataFrame(scores,            
            columns=['國文', '英文', '數學', '地理', '歷史'],
            index=range(1,51))
##scores_df.T
names_df:DataFrame = pd.read_csv('students11.csv')
names_df['姓名'].head(n=50).values
scores_df[['姓名','性別']] = names_df[['姓名','性別']].head(n=50).values
scores_df = scores_df[['姓名', '性別', '國文', '英文', '數學', '地理', '歷史']]

#統計
scores_df['總分'] = scores_df[['國文', '英文', '數學', '地理', '歷史']].sum(axis=1)       # 加總
scores_df['平均'] = scores_df[['國文', '英文', '數學', '地理', '歷史']].mean(axis=1)       # 平均
scores_df['排名'] = scores_df[['總分']].rank(ascending=False,method='min')       # 排名
scores_df.sort_values(by='排名') #排序
scores_df
```
