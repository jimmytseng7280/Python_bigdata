```python
#DataFrmae(ndArray(無欄名列名，只有數據)升級版加了欄名列名)
import numpy as np
from numpy import ndarray
import pandas as pd
from pandas import DataFrame

scores:ndarray | None = None #學生分數
scores_df:DataFrame | None = None #學生分數的DataFrame

scores = np.random.randint(50, 101, size=(50, 5))
#變數命名法:有一種scores_df, 也可以studentScoresDf
scores_df = pd.DataFrame(scores, 
             index=range(1,51),
             columns=['國文','英文','數學','地理','歷史'])

scores_df.T #樞紐分析:是以X軸Y軸互換做分析
```
