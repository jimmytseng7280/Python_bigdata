```python
#建立工廠 (最前面第一個英文要大寫，初始化init()建立實體)

class Person:
    #pass #弱勢pass是指程式直接跳過，所以會為空。
    def __init__(self): #自訂的初始化
        print("初始化:一個新的人被建立了")
```

```python
p1 = Person()
type(p1)
```

```python
class Person:
    def __init__(self,name):
        #建立一個attribute
        self.name = name

    #自訂實體的method(方法)
    def introduce(self):
        print(f"哈囉，我的名字是{self.name}")
```

```python
p2 = Person("robert")
p2.name
```

```python
p3 = Person("jenny")
p3.name
p2.introduce()
```
