```python
# 載入數值運算套件
import numpy as np

# 載入繪圖套件
import matplotlib.pyplot as plt

# ============================
# 設定中文字型
# ============================

# Windows 常用中文字型：微軟正黑體
plt.rcParams['font.sans-serif'] = ['Microsoft JhengHei']

# 避免負號顯示成方塊
plt.rcParams['axes.unicode_minus'] = False

# ============================
# 建立 x 軸資料
# ============================

# 產生 -10 到 10 的 1000 個點
x = np.linspace(-10, 10, 1000)

# ============================
# 定義各種活化函數
# ============================

# Sigmoid 函數
# 公式：1 / (1 + e^(-x))
sigmoid = 1 / (1 + np.exp(-x))

# Tanh（雙曲正切）函數
# 輸出範圍：-1 ~ 1
tanh = np.tanh(x)

# ReLU（修正線性單元）
# x > 0 時輸出 x
# x <= 0 時輸出 0
relu = np.maximum(0, x)

# Leaky ReLU（洩漏式 ReLU）
# x <= 0 時保留 1% 斜率
leaky_relu = np.where(x > 0, x, 0.01 * x)

# ELU（指數線性單元）
# x > 0 時輸出 x
# x <= 0 時輸出 e^x - 1
elu = np.where(x > 0, x, np.exp(x) - 1)

# Swish 函數
# Google 提出的活化函數
swish = x / (1 + np.exp(-x))

# ============================
# 建立畫布
# ============================

# 圖片大小：12吋 × 8吋
plt.figure(figsize=(12, 8))

# ============================
# 第一張圖：Sigmoid S型活化函數
# ============================

plt.subplot(2, 3, 1)      # 2列3欄，第1張
plt.plot(x, sigmoid)
plt.title("S型活化函數（Sigmoid）")
plt.xlabel("輸入值 x")
plt.ylabel("輸出值")
plt.grid(True)

# ============================
# 第二張圖：Tanh 雙曲正切函數
# ============================

plt.subplot(2, 3, 2)
plt.plot(x, tanh)
plt.title("雙曲正切函數（Tanh）")
plt.xlabel("輸入值 x")
plt.ylabel("輸出值")
plt.grid(True)

# ============================
# 第三張圖：ReLU 修正線性單元
# ============================

plt.subplot(2, 3, 3)
plt.plot(x, relu)
plt.title("修正線性單元（ReLU）")
plt.xlabel("輸入值 x")
plt.ylabel("輸出值")
plt.grid(True)

# ============================
# 第四張圖：Leaky ReLU
# ============================

plt.subplot(2, 3, 4)
plt.plot(x, leaky_relu)
plt.title("洩漏式修正線性單元")
plt.xlabel("輸入值 x")
plt.ylabel("輸出值")
plt.grid(True)

# ============================
# 第五張圖：ELU
# ============================

plt.subplot(2, 3, 5)
plt.plot(x, elu)
plt.title("指數線性單元（ELU）")
plt.xlabel("輸入值 x")
plt.ylabel("輸出值")
plt.grid(True)

# ============================
# 第六張圖：Swish
# ============================

plt.subplot(2, 3, 6)
plt.plot(x, swish)
plt.title("自門控活化函數（Swish）")
plt.xlabel("輸入值 x")
plt.ylabel("輸出值")
plt.grid(True)

# ============================
# 自動調整子圖間距
# ============================

plt.tight_layout()

# ============================
# 顯示圖形
# ============================

plt.show()
```
