```python
#標準寫法
import numpy as np
import matplotlib.pyplot as plt

# 製作figure
fig = plt.figure()

# 在figure上製作一個axes
ax_1 = fig.add_subplot(1, 1, 1)

# 產生 x 軸資料 (0 ~ 2π)
x = np.linspace(0, 2 * np.pi, 1000)
ax_0 = np.sin(x)

# 計算正弦波
y = np.sin(x)

# 繪圖
plt.plot(x, y)

# 設定標題與座標軸名稱
plt.title("Sine Wave")
plt.xlabel("x")
plt.ylabel("sin(x)")

# 顯示格線
plt.grid(True)

# 顯示圖
plt.show()
```
