```python
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['Microsoft JhengHei', 'SimHei', 'Arial Unicode MS']
plt.rcParams['axes.unicode_minus'] = False

# 讀取 CSV
df = pd.read_csv("考試分數_3年6班.csv")

# 篩選趙冠宇的資料
zhao = df[df["學生姓名"] == "趙冠宇"]

# 取出各科成績
scores = zhao.iloc[0][["語文", "數學", "英語", "物理", "化學"]]

# 繪製長條圖
fig = plt.figure(figsize=(6, 4))
ax = fig.add_subplot(1, 1, 1)

ax.bar(scores.index, scores.values)

ax.set_title("趙冠宇成績")
ax.set_xlabel("科目")
ax.set_ylabel("分數")
ax.set_ylim(0, 100)
ax.set_xticks(range(len(scores.index)))
ax.set_xticklabels(scores.index)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['Microsoft JhengHei', 'SimHei', 'Arial Unicode MS']
plt.rcParams['axes.unicode_minus'] = False

df = pd.read_csv("考試分數_3年6班.csv")
search_name = df[df["學生姓名"] == "趙冠宇"]

if search_name.empty:
    print("找不到這位學生")
else:
    scores = search_name.iloc[0][["語文", "數學", "英語", "物理", "化學"]]

    fig, (ax1, ax2) = plt.subplots(
        2, 1, figsize=(8, 8),
        gridspec_kw={'height_ratios': [3, 1]}
    )

    ax1.bar(scores.index, scores.values)
    ax1.set_title("趙冠宇各科成績")
    ax1.set_xlabel("科目")
    ax1.set_ylabel("分數")
    ax1.set_ylim(0, 100)

    ax2.axis("off")
    table_data = [[score for score in scores.values]]
    table = ax2.table(
        cellText=table_data,
        colLabels=scores.index,
        rowLabels=["趙冠宇"],
        loc="center",
        cellLoc="center"
    )
    table.auto_set_font_size(False)
    table.set_fontsize(12)
    table.scale(1, 1.5)

    plt.tight_layout()
    plt.show()
```

```python
#自行輸入學生姓名，查詢成績並以圖表表示
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['Microsoft JhengHei', 'SimHei', 'Arial Unicode MS']
plt.rcParams['axes.unicode_minus'] = False

df = pd.read_csv("考試分數_3年6班.csv")

name = input("請輸入學生姓名：")
search_name = df[df["學生姓名"] == name]

if search_name.empty:
    print("找不到這位學生")
else:
    scores = search_name.iloc[0][["語文", "數學", "英語", "物理", "化學"]]

    fig, (ax1, ax2) = plt.subplots(
        2, 1, figsize=(8, 8),
        gridspec_kw={'height_ratios': [3, 1]}
    )

    ax1.bar(scores.index, scores.values)
    ax1.set_title(f"{name} 各科成績")
    ax1.set_xlabel("科目")
    ax1.set_ylabel("分數")
    ax1.set_ylim(0, 100)

    ax2.axis("off")
    table_data = [[score for score in scores.values]]
    table = ax2.table(
        cellText=table_data,
        colLabels=scores.index,
        rowLabels=[name],
        loc="center",
        cellLoc="center"
    )
    table.auto_set_font_size(False)
    table.set_fontsize(12)
    table.scale(1, 1.5)

    plt.tight_layout()
    plt.show()
```
