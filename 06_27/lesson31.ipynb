```python
import pandas as pd
import numpy as np

# 1. 讀取 CSV 檔
dataFrame = pd.read_csv('各班圖書借閱統計.csv')

# 2. 第 0 列(index 為 0)才是真正的中文欄位名稱, 把它設為欄位名
dataFrame.columns = dataFrame.loc[0]

# 3. 刪除第 0 列(已經拿來當欄位名, 資料中不需要重複)
dataFrame = dataFrame.drop(index=0)

# 4. 取得目前的列數與欄數（在刪除尾端列之前先獲取總行數）
rowCount, columnCount = dataFrame.shape

# 5. 先印出 dataFrame 看看那些列的 index 是多少
print("--- 刪除前的原始資料 ---")
print(dataFrame)

# 【正確完成】：刪除合計列與 4 列說明文字（共 5 列）
# 利用 rowCount 往回推算最後 5 列的 index 編號
dataFrame.drop(range(rowCount - 4, rowCount + 1), inplace=True)


# 【正確完成】：把索引重設為從 0 開始的連續編號
# 利用新的資料列數，生成一個從 0 開始的連續數字範圍
dataFrame.index = range(len(dataFrame))


# 【正確完成】：只保留「班級」「學生人數」「借閱冊數」三個欄位
dataFrame1 = dataFrame.reindex(columns=['班級', '學生人數', '借閱冊數'])


# 6. 印出最終清洗完畢的結果
print("\n--- 清洗後的最終結果 ---")
print(dataFrame1)
```
