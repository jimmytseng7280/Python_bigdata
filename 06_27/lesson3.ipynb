```python
import pandas as pd
dataFrame = pd.read_csv('各鄉鎮市區人口密度.csv')

dataFrame.columns = ['統計年', '區域別', '年底人口數', '土地面積', '人口密度'] #改變欄位名稱
df = dataFrame.drop(index=0) #刪除第一行
df1 = df.dropna() #刪除缺失值
df2 = df1.drop('統計年', axis=1) #刪除'統計年'欄位
df4 = pd.to_numeric(dataFrame['年底人口數'], errors='coerce') # 將'年底人口數'欄位轉換為數值型態，無法轉換的值將被設為NaN
df5 = df4.dropna()

df3 = df2.copy() # 建立 df2 的副本，避免直接修改原始資料
df3['年底人口數'] = pd.to_numeric(df3['年底人口數'], errors='coerce') # 安全轉換法：轉換成數字，如果遇到無法轉換的文字會先變成 NaN
df3['年底人口數'] = df3['年底人口數'].astype(int) # 確保轉為整數型態 (int)
print(df3['年底人口數'].dtype) # 檢查轉換後的資料型態（會顯示 int64）

df3
```
