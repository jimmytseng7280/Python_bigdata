```python
#1) list：要自己一個個處理。list 適合一般資料儲存。

a = [1, 2, 3]
print(a + [4, 5, 6])  #[1, 2, 3, 4, 5, 6]
```

```python
#2) numpy：可以直接整個做運算。numpy 適合數值運算，寫法更簡潔，也通常更快。

import numpy as np

b = np.array([1, 2, 3])
print(b + np.array([4, 5, 6]))  #[5 7 9]
```

```python
#numpy範例-陣列運算

import numpy as np #將numpy命名(小名)為np
arr1 = np.array([6, 7.5, 8, 0]) #建立一個一維陣列 ndarray
print(arr1) #印出陣列

arr2 = arr1 * 5
print(arr2)

arr3 = arr1 / 10
print(arr3)
```

```python
import numpy as np #將numpy命名(小名)為np
np.arange(5) #建立一個一維陣列，從1開始到10結束，間隔為2
np.arange(2, 6) #建立一個一維陣列，從1開始到10結束，間隔為2
np.arange(1, 10, 2) #建立一個一維陣列，從1開始到10結束，間隔為2
```

```python
import numpy as np
import math
import matplotlib.pyplot as plt

t = np.arange(0, 2.5, 0.1)
print(len(t)) #查個數
t1 = t * math.pi
y1 = np.sin(t1)
y2 = np.cos(t1)
print(len(y1)) #查個數
print(len(y2)) #查個數

fig = plt.figure()
axes = fig.add_subplot(1,1,1)
axes.plot(t, y1)
axes.plot(t, y2)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.arange(0, 5, 0.1)
y1 = np.sin(t * np.pi)
y2 = np.cos(t * np.pi)
figure = plt.figure()
ax1 = figure.add_subplot(2,1,1)
ax1.plot(t, y1)
ax2 = figure.add_subplot(2,1,2)
ax2.plot(t, y2)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.arange(0, 5, 0.1)
y1 = np.sin(t * np.pi)
y2 = np.cos(t * np.pi) 

figure = plt.figure()
ax1 = figure.add_subplot(2, 1, 1)
ax1.plot(t, y1, linestyle="--", color="red")
ax2 = figure.add_subplot(2, 1, 2)
ax2.plot(t, y2, linestyle="-.",color="blue")
plt.show()
```
