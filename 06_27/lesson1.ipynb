```python
import numpy as np
import math
import matplotlib.pyplot as plt

t = np.arange(0, 2.5, 0.1)
print(len(t)) #查個數
t1 = t * math.pi
y1 = np.sin(t1)
y2 = np.cos(t1)
print(len(y1)) #查個數
print(len(y2)) #查個數

fig = plt.figure()
axes = fig.add_subplot(1,1,1)
axes.plot(t, y1)
axes.plot(t, y2)
plt.show()
```

```python
#純註解半成品(依註解寫程式)
#請依照下方每一行註解的說明,寫出對應的 Python 程式碼。

# 步驟 1:匯入 numpy,並取別名 np
import numpy as np

# 步驟 2:匯入 matplotlib.pyplot,並取別名 plt
import matplotlib.pyplot as plt

# 步驟 3:用 np.arange() 產生 0 到 4(不含 4)、間隔 0.2 的陣列,存入 x
x = np.arange(0, 4, 0.2)

# 步驟 4:計算 y1 = cos(pi * x)(用 np.cos 與 np.pi)
y1 = np.cos(np.pi * x)

# 步驟 5:計算 y2 = cos(2 * pi * x)
y2 = np.cos(2 * np.pi * x)

# 步驟 6:用 subplot 把版面切成 1 列 2 欄,選第 1 格(左圖)
figure = plt.figure()
ax1 = figure.add_subplot(1, 2, 1)

# 步驟 7:畫出 x 與 y1,使用綠色實線
ax1.plot(x ,y1, linestyle="-", color="green")

# 步驟 8:用 subplot 選第 2 格(右圖)
ax2 = fig.add_subplot(1, 2, 2)

# 步驟 9:畫出 x 與 y2,使用洋紅色點線
ax2.plot(x ,y2, linestyle="-.", color="red")

# 步驟 10:顯示圖形
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 4, 0.2)
y1 = np.cos(np.pi * x)
y2 = np.cos(2 * np.pi * x)

plt.subplot(1, 2, 1)
plt.plot(x, y1, 'g-')

plt.subplot(1, 2, 2)
plt.plot(x, y2, 'm:')

plt.show()
```
